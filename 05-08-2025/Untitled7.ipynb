```python
import csv
data = []

with open("Finds.csv", "r") as file:
    reader = csv.reader(file)
    next(reader)
    for row in reader:
        data.append(row)

h = data[0][:-1]

for row in data:
    if row[-1] == "Yes":
        for i in range(len(h)):
            if h[i] != row[i]:
                h[i] = "?"

print("Final Hypothesis:")
print(h)
```

```python
import csv
data = []

with open("Finds.csv", "r") as file:
    reader = csv.reader(file)

    next(reader)      

    for row in reader:
        data.append(row)

num_attr = len(data[0]) - 1

S = data[0][:-1]

G = [['?' for i in range(num_attr)] for j in range(num_attr)]

print("Initial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
for g in G:
    print(g)
for row in data:

    if row[-1] == "Yes":

        for i in range(num_attr):

            if row[i] != S[i]:
                S[i] = '?'
                G[i][i] = '?'
    else:

        for i in range(num_attr):

            if row[i] != S[i]:
                G[i][i] = S[i]
            else:
                G[i][i] = '?'

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")

for g in G:
    if g != ['?'] * num_attr:
        print(g)
```
