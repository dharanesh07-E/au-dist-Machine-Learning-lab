```python
import pandas as pd
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
from sklearn.metrics import classification_report
from sklearn.metrics import accuracy_score

df = pd.read_csv("iris1.csv")
print(df.head(10))

array = df.values
X = array[:, 0:4]
y = array[:, 4]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=0, stratify=y
)

model = SVC(kernel='linear', random_state=0)
model.fit(X_train, y_train)

pred = model.predict(X_test)

print("\nThe confusion matrix is\n")
conf = confusion_matrix(y_test, pred)
print(conf)

print("\nAccuracy is")
accuracy = accuracy_score(y_test, pred)
print(accuracy)

print("\nClassification Report")
report = classification_report(y_test, pred)
print(report)

model1 = SVC(kernel='rbf', random_state=0)
model1.fit(X_train, y_train)

pred1 = model1.predict(X_test)

print("\nThe confusion matrix for RBF kernel is\n")
conf1 = confusion_matrix(y_test, pred1)
print(conf1)

print("\nAccuracy for RBF kernel is")
accuracy1 = accuracy_score(y_test, pred1)
print(accuracy1)

print("\nClassification Report for RBF kernel")
report1 = classification_report(y_test, pred1)
print(report1)
```
