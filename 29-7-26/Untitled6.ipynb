```python
import numpy as np
from sklearn.decomposition import PCA
x=np.array([[2,1],[3,2],[4,3]])
pca=PCA(n_components=1)
principal_component=pca.fit_transform(x)
print("principal component")
print(principal_component)
print("\n eign value")
print(pca.explained_variance_)
print("\n eign vector")
print(pca.components_)
```

```python
import numpy as np
A=np.array([[2,0],[0,1]])
U,S,VT=np.linalg.svd(A)
print("U matrix")
print(U)
print("\n Singular values")
print(S)
print("\n Sigma Matrix")
sigma=np.zeros((2,2))
sigma[:2,:2]=np.diag(S);
print(sigma)
print("\n transpose");
print(VT)
A_reconstruct=U@sigma@VT
print("\n Reconstructed MATRIX")
print(A_reconstruct)

```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression


X = np.array([[1], [2], [3], [4], [5]])
Y = np.array([2, 4, 6, 8, 10])

model = LinearRegression()


model.fit(X, Y)


coefficient = model.coef_
intercept = model.intercept_
prediction = model.predict([[6]])

y_pred = model.predict(X)


score = model.score(X, Y)

print("========== Linear Regression ==========\n")

print("Coefficient (Slope):", coefficient)
print("Intercept:", intercept)

print("\nRegression Equation")
print(f"Y = {intercept:.2f} + ({coefficient[0]:.2f})X")

print("\nPrediction for X = 6")
print(prediction)

print("\nR² Score")
print(score)

print("\nActual Values")
print(Y)

print("\nPredicted Values")
print(y_pred)

plt.figure(figsize=(7,5))

plt.scatter(X, Y,
            color='red',
            label='Actual Data')

plt.plot(X,
         y_pred,
         color='blue',
         linewidth=2,
         label='Regression Line')

plt.title("Linear Regression")
plt.xlabel("X")
plt.ylabel("Y")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

X = np.array([
    [1,22],
    [2,24],
    [3,25],
    [4,27],
    [5,30]
])

Y = np.array([
    25000,
    30000,
    40000,
    50000,
    65000
])

model = LinearRegression()

model.fit(X, Y)

coefficients = model.coef_
intercept = model.intercept_

prediction = model.predict([[6,32]])

y_pred = model.predict(X)

score = model.score(X,Y)

print("========== Multiple Linear Regression ==========\n")

print("Coefficient for Experience :", coefficients[0])
print("Coefficient for Age        :", coefficients[1])

print("\nIntercept :", intercept)

print("\nRegression Equation")

print(f"Y = {intercept:.2f} + ({coefficients[0]:.2f})X1 + ({coefficients[1]:.2f})X2")

print("\nPrediction for")
print("Experience = 6")
print("Age = 32")

print(prediction)

print("\nR² Score")
print(score)

print("\nActual Salary")
print(Y)

print("\nPredicted Salary")
print(y_pred)

plt.figure(figsize=(7,5))

plt.scatter(Y,
            y_pred,
            color='blue',
            label='Predicted')

plt.plot([min(Y), max(Y)],
         [min(Y), max(Y)],
         color='red',
         linewidth=2,
         label='Ideal Line')

plt.title("Multiple Linear Regression")
plt.xlabel("Actual Salary")
plt.ylabel("Predicted Salary")

plt.legend()
plt.grid(True)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import PolynomialFeatures
from sklearn.linear_model import LinearRegression
x=np.array([[1],[2],[3],[4],[5]]);
y=np.array([1,4,9,16,25])
poly=PolynomialFeatures(degree=2)
x_poly=poly.fit_transform(x)
model=LinearRegression()
model.fit(x_poly,y)
x_grid=np.arange(1,5.1,0.1).reshape(-1,1)
plt.scatter(x,y,color='red')
plt.plot(x_grid,model.predict(poly.transform(x_grid)),color='blue')
plt.title("polynomial regression")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
print("coefficient:",model.coef_)
print("Intercept:",model.intercept_)
print("predcition:",model.predict(poly.transform([[6]])))
print("R2 score:",model.score(x_poly,y))     
```

```python
import csv
data = []

with open("Finds.csv", "r") as file:
    reader = csv.reader(file)

    next(reader)      

    for row in reader:
        data.append(row)

num_attr = len(data[0]) - 1

S = data[0][:-1]

G = [['?' for i in range(num_attr)] for j in range(num_attr)]

print("Initial Specific Hypothesis:")
print(S)

print("\nInitial General Hypothesis:")
for g in G:
    print(g)
for row in data:

    if row[-1] == "Yes":

        for i in range(num_attr):

            if row[i] != S[i]:
                S[i] = '?'
                G[i][i] = '?'
    else:

        for i in range(num_attr):

            if row[i] != S[i]:
                G[i][i] = S[i]
            else:
                G[i][i] = '?'

print("\nFinal Specific Hypothesis:")
print(S)

print("\nFinal General Hypothesis:")

for g in G:
    if g != ['?'] * num_attr:
        print(g)
```
